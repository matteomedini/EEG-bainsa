```python
from pathlib import Path

import numpy as np
import pandas as pd

from moabb.datasets import BNCI2014_001
from moabb.paradigms import MotorImagery
```

```python
dataset = BNCI2014_001()

paradigm = MotorImagery(
    n_classes=4,
    fmin=8,
    fmax=32,
    tmin=0.0,
    tmax=4.0,
)

X, y, metadata = paradigm.get_data(dataset=dataset)

print("X shape:", X.shape)
print("y shape:", y.shape)
print("metadata shape:", metadata.shape)
print("labels:", sorted(pd.Series(y).unique().tolist()))
print(metadata.head())
```

```python
print(metadata.columns.tolist())
print(metadata["subject"].value_counts().sort_index())
print(metadata["session"].value_counts().sort_index())
print(metadata["run"].value_counts().sort_index())
print(pd.Series(y).value_counts())
print("Shape singola prova:", X[0].shape)
```

```python
out_dir = Path("prepared_bnci2014_001")
out_dir.mkdir(exist_ok=True)

np.save(out_dir / "X.npy", X)
np.save(out_dir / "y.npy", y)
metadata.to_csv(out_dir / "metadata.csv", index=False)

print("Saved:")
print(out_dir / "X.npy")
print(out_dir / "y.npy")
print(out_dir / "metadata.csv")
```

```python
X_check = np.load(out_dir / "X.npy", allow_pickle=True)
y_check = np.load(out_dir / "y.npy", allow_pickle=True)
meta_check = pd.read_csv(out_dir / "metadata.csv")

print(X_check.shape)
print(y_check.shape)
print(meta_check.shape)
print(meta_check.head())
```
